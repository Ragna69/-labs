```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
# загрузка датасета House Prices
try:
    df = pd.read_csv('train.csv')
    print("Датасет успешно загружен!")
except FileNotFoundError:
    print("Ошибка: Файл 'train.csv' не найден. Пожалуйста, скачайте его с Kaggle.")
```

```python
# вывод первых 5 строк
print(" Первые 5 строк датасета ")
display(df.head())

# общая информация о типах данных и пропусках
print("\n Общая информация ")
df.info()
```

```python
# получение количества пропущенных значений для каждого столбца
print(" Количество пропущенных значений ")
missing_values = df.isnull().sum()

# аыводит только те столбцы, где есть пропуски, чтобы не загромождать вывод
print(missing_values[missing_values > 0])
```

```python
print(" Заполнение пропусков ")

# находит все столбцы, где есть хотя бы один пропуск
cols_with_missing = df.columns[df.isnull().any()].tolist()

for col in cols_with_missing:
    if df[col].dtype == 'object':
        # для категориальных признаков используем МОДУ
        mode_val = df[col].mode()[0]
        df[col].fillna(mode_val, inplace=True)
        print(f"Столбец '{col}' заполнен модой: {mode_val}")
    else:
        # для числовых признаков используем МЕДИАНУ
        median_val = df[col].median()
        df[col].fillna(median_val, inplace=True)
        print(f"Столбец '{col}' заполнен медианой: {median_val}")

# проверка: сумма всех пропусков должна быть равна 0
print("\nПроверка заполнения (должно быть 0):", df.isnull().sum().sum())
```

```python
print(" Нормализация данных (MinMaxScaler) ")

# аыбирает числовые столбцы для нормализации
numeric_cols = ['LotArea', 'GrLivArea', 'TotalBsmtSF', 'GarageCars', 'YearBuilt']

# фильтрует список, оставляя только существующие столбцы
existing_numeric = [col for col in numeric_cols if col in df.columns]

if existing_numeric:
    scaler = MinMaxScaler()
    # fit_transform вычисляет параметры и сразу преобразует данные в диапазон [0, 1]
    df[existing_numeric] = scaler.fit_transform(df[existing_numeric])
    
    print(f"Нормализованы столбцы: {existing_numeric}")
    display(df[existing_numeric].head())
else:
    print("Нет указанных числовых столбцов для нормализации.")
```

```python
print(" One-Hot Encoding категориальных данных ")

# находит все текстовые столбцы
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

if categorical_cols:
    # drop_first=True удаляет первый столбец категории, чтобы избежать мультиколлинеарности
    df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
    
    print(f"Размер до кодирования: {df.shape}")
    print(f"Размер после кодирования: {df_encoded.shape}")
    
    print("\nПервые 5 строк итогового датасета:")
    display(df_encoded.head())
    
    # сохраняет итоговый датафрейм
    df_final = df_encoded
else:
    df_final = df
```
